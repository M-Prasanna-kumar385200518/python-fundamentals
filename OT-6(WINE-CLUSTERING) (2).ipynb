```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\madha\Downloads\wine-clustering (1).csv")

def remove_outliers(dataframe):
    Q1 = dataframe.quantile(0.25)
    Q3 = dataframe.quantile(0.75)
    IQR = Q3 - Q1

    df_clean = dataframe[~((dataframe < (Q1 - 1.5 * IQR)) | (dataframe > (Q3 + 1.5 * IQR))).any(axis=1)]
    return df_clean

df_clean = remove_outliers(df)

plt.figure(figsize=(14, 6))
features = df_clean.columns

fig, axes = plt.subplots(nrows=3, ncols=5, figsize=(15, 8))
fig.suptitle('Boxplots of Wine Features (Without Outliers)', fontsize=14, fontweight='bold', color='darkred')

axes = axes.flatten()

for i, feature in enumerate(features):
    sns.boxplot(y=df_clean[feature], ax=axes[i], color="teal")
    axes[i].set_title(feature, fontsize=12, fontweight='bold', color='navy')

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import mahalanobis

df = pd.read_csv(r"C:\Users\madha\Downloads\wine-clustering (1).csv")

def mahalanobis_distance(x, data):
    cov_matrix = np.cov(data.T)
    inv_cov_matrix = np.linalg.inv(cov_matrix)
    mean = np.mean(data, axis=0)
    return np.array([mahalanobis(row, mean, inv_cov_matrix) for row in x])

feature1, feature2 = df.columns[0], df.columns[1]
data_subset = df[[feature1, feature2]].values

distances = mahalanobis_distance(data_subset, data_subset)
threshold = np.percentile(distances, 97.5)  # Set threshold (e.g., top 2.5% as outliers)

df_clean = df[distances < threshold]

df_clean.shape
```
